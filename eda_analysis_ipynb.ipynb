```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# 1. Load the scraped dataset
df = pd.read_csv("codealpha_scraped_books.csv")

# Display first few rows and basic info
print("--- Dataset Info ---")
print(df.info())
print("\n--- First 5 Rows ---")
display(df.head())

# 2. Clean the Price column (remove '£' and convert to float)
df['Price_Clean'] = df['Price'].str.replace('£', '', regex=False).astype(float)

# 3. Display summary statistics
print("\n--- Summary Statistics ---")
display(df['Price_Clean'].describe())
```

```python
# 1. Price Distribution Histogram with KDE curve
plt.figure(figsize=(8, 5))
sns.histplot(df['Price_Clean'], kde=True, color='royalblue', bins=15)
plt.title('Distribution of Book Prices')
plt.xlabel('Price (£)')
plt.ylabel('Frequency')
plt.show()

# 2. Box Plot for Outlier Detection
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['Price_Clean'], color='orange')
plt.title('Book Price Boxplot (Outlier Detection)')
plt.xlabel('Price (£)')
plt.show()
```
